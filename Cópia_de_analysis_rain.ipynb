```python
# Inserir dados de Entrada
file_path = '/content/drive/My Drive/HIDROLOGIA/datasets_postos/637032_Chuvas.csv'
pluv1 = pd.read_csv(file_path, encoding='utf-8', sep=';',decimal=',')
pluv1.head()
```

```python
#Avaliação dos dados
for i, col in enumerate(pluv1.columns):
    print(f'Índice {i}: {col}')
```

```python
#Excluindo colunas Desnecessárias
pluv1 = pluv1.drop(pluv1.columns[6:75], axis=1)
pluv1 = pluv1.drop(pluv1.columns[[1, 3, 4]], axis=1)
display(pluv1)
```

```python
# Selecionando os registros entre 1940 e 1970 e excluindo o resto
pluv1["Data"] = pd.to_datetime(pluv1["Data"], dayfirst=True)
pluv1 = pluv1[(pluv1["Data"].dt.year >= 1941) & (pluv1["Data"].dt.year <= 1970)]

# Exibindo o resultado final
display(pluv1)
```

```python
# Adicionando meses faltantes e Corrigindo
pluv1["Data"] = pd.to_datetime(pluv1["Data"], dayfirst=True)
#display(pluv1)
meses_presentes = pluv1["Data"].dt.to_period("M").unique()
df_meses = pd.DataFrame({"Meses Presentes": meses_presentes})
#display(df_meses)
todos_os_meses = pd.period_range(start=pluv1["Data"].min(),
                                 end=pluv1["Data"].max(),
                                 freq="M")

df = pd.DataFrame({"Meses": todos_os_meses})
meses_faltando = [mes for mes in todos_os_meses if mes not in meses_presentes]
#display(df)
# Criar um DataFrame **somente se houver meses faltando**
if meses_faltando:
    df_faltantes = pd.DataFrame({"Meses Faltando": meses_faltando})
    display(df_faltantes)
else:
    print("Nenhum mês está faltando!")
```

```python
total_meses = len(todos_os_meses)
quantidade_faltante = len(meses_faltando)
porcentagem_erro = (quantidade_faltante / total_meses) * 100
print(f"Total de meses: {total_meses}")
print(f"Quantidade de meses faltando: {quantidade_faltante}")
print(f"Porcentagem de erro: {porcentagem_erro:.2f} %")
```

```python
#incluir mês faltantes ao data frame

df_todos_meses = pd.DataFrame({"Data": pd.date_range(start=pluv1["Data"].min(),
                                                     end=pluv1["Data"].max(),
                                                     freq="MS")})
# Fazendo um merge para garantir que todos os meses estejam no DataFrame original
pluv1 = df_todos_meses.merge(pluv1, on="Data", how="left")

# Preenchendo os valores ausentes com -99
pluv1['Total']= pluv1['Total'].fillna(-99)
pluv1["EstacaoCodigo"] = pluv1["EstacaoCodigo"].fillna('637032')

# Exibindo o resultado final
display(pluv1)

```

Todo o script acima é sobre limpar os dados brutos e verificação de meses ausente ao data frame , além é possível gerar uma planilha com a adição dos mese ausente e um fator de erro associado(-99)

```python


# Criar coluna de Ano e Mês
pluv1["Ano"] = pluv1["Data"].dt.year
pluv1["Mes"] = pluv1["Data"].dt.month
pluv1["Total"] = pd.to_numeric(pluv1["Total"], errors="coerce")


# Função para calcular a soma por ano e definir -99 caso falte algum mês
def soma_anual(df):
    # Agrupar por Ano e calcular a soma, ignorando NaNs
    soma_total = df.groupby("Ano")["Total"].sum(min_count=1)
    contagem_meses = df.groupby("Ano")["Mes"].nunique()

    # Se algum mês estiver faltando, define a soma como -99
    soma_total[contagem_meses < 12] = -99
    soma_total = soma_total.fillna(-99)

    df_soma_anual = soma_total.reset_index()
    df_soma_anual.columns = ["Ano", "Soma Anual"]

    return df_soma_anual

# Chamar a função e exibir o resultado
dt_soma_anual = soma_anual(pluv1)
dt_soma_anual['Soma Anual'] = dt_soma_anual['Soma Anual'].round(2)
display(dt_soma_anual)
```
